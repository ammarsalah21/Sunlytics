```python
import pandas as pd

# Assuming your merged and cleaned data is in "cleaned_df"
# (if it's called merged_df instead, just replace it below)

# 1️⃣ Main area — Growth & Urbanization
main_df = cleaned_df[[
    "Country",
    "Year",
    "Solar capacity (total) (GW)",
    "World Bank's income classification",
    "Urban population"
]].copy()

# 2️⃣ Efficiency area — Economic & Operational Efficiency
efficiency_df = cleaned_df[[
    "Country",
    "Year",
    
    "Solar capacity (total) (GW)",
    "Solar per capita (kWh - equivalent)",
    "GDP per capita",
    "Electricity from solar - TWh"
]].copy()

# 3️⃣ Environment area — Environmental Impact & Policies
environment_df = cleaned_df[[
    "Country",
    "Year",
    "Solar (TWh growth - equivalent)",
    "Annual CO₂ emissions (per capita)",
    "Electricity from solar (TWh)",
    "Solar (% equivalent primary energy)"
]].copy()

# 4️⃣ Save all three to Excel files
main_df.to_excel("Main_Data.xlsx", index=False)
efficiency_df.to_excel("Efficiency_Data.xlsx", index=False)
environment_df.to_excel("Environment_Data.xlsx", index=False)

print("✅ All DataFrames created and saved successfully!")
```

```python
# 1️⃣ Define a function to clean the DataFrames
def clean_table(df):
    exclude_cols = ["Country", "Year"]
    other_cols = [col for col in df.columns if col not in exclude_cols]
    return df.dropna(subset=other_cols, how="all").reset_index(drop=True)

# 2️⃣ Apply cleaning to all three tables
main_df = clean_table(main_df)
efficiency_df = clean_table(efficiency_df)
environment_df = clean_table(environment_df)

# 3️⃣ Save them all into a NEW Excel file
with pd.ExcelWriter("Cleaned_Solar_Study_Areas.xlsx") as writer:
    main_df.to_excel(writer, sheet_name="Main_Area", index=False)
    efficiency_df.to_excel(writer, sheet_name="Efficiency_Area", index=False)
    environment_df.to_excel(writer, sheet_name="Environment_Area", index=False)

print("✅ Cleaned data saved successfully in 'Cleaned_Solar_Study_Areas.xlsx'")
```

```python
import pandas as pd

 
file_path = r"c:\Users\alhaiaah\Downloads\Sol.xlsx"   

import pandas as pd

file_path = r"your_file.xlsx"

dimYear = pd.read_excel(file_path, sheet_name="dimYear")
dimCountry = pd.read_excel(file_path, sheet_name="dimCountry")
dimIncome = pd.read_excel(file_path, sheet_name="dimIncome")   # الاسم صحيح
factSolar = pd.read_excel(file_path, sheet_name="factSolar ")  # ← لاحظ الـ Space هنا

# --- mapping ---
country_map = dict(zip(dimCountry["Country"], dimCountry["CountryID"]))
year_map = dict(zip(dimYear["Year"], dimYear["YearID"]))
income_map = dict(zip(dimIncome["Income"], dimIncome["IncomeID"]))

# --- replace text with IDs ---
factSolar["CountryID"] = factSolar["Country"].map(country_map)
factSolar["YearID"] = factSolar["Year"].map(year_map)
factSolar["IncomeID"] = factSolar["Income"].map(income_map)

# remove old text columns
fact_final = factSolar.drop(columns=["Country", "Year", "Income"])

# reorder
id_cols = ["CountryID", "YearID", "IncomeID"]
other_cols = [c for c in fact_final.columns if c not in id_cols]
fact_final = fact_final[id_cols + other_cols]

# save
output_path = r"factSolar_cleaned.xlsx"
fact_final.to_excel(output_path, index=False)

 
```
